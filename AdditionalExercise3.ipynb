```python
from pathlib import Path
import csv
import json
import shutil
from collections import Counter

# Student Value
SEED_DIGIT = 6

# Target Year
TARGET_YEAR = 2015 + SEED_DIGIT

# Folders
data_dir = Path("data")
output_dir = Path("output")
backup_dir = Path("backup")

output_dir.mkdir(exist_ok=True)
backup_dir.mkdir(exist_ok=True)

# Dataset
dataset_path = data_dir / "Global_Cybersecurity_Threats_2015-2024.csv"

records = []
attack_counter = Counter()

total_records = 0
incident_count = 0
total_loss = 0

with open(dataset_path, "r", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    for row in reader:

        total_records += 1

        if int(row["Year"]) == TARGET_YEAR:

            loss = float(row["Financial Loss (in Million $)"])

            records.append(row)

            attack_counter[row["Attack Type"]] += 1

            incident_count += 1
            total_loss += loss

# Average Loss
if incident_count > 0:
    average_loss = total_loss / incident_count
else:
    average_loss = 0

# Five Costliest Incidents
top_incidents = sorted(
    records,
    key=lambda x: float(x["Financial Loss (in Million $)"]),
    reverse=True
)[:5]

# Export Filtered CSV
csv_file = output_dir / "cybersecurity_2021.csv"

with open(csv_file, "w", newline="", encoding="utf-8") as file:

    writer = csv.DictWriter(
        file,
        fieldnames=reader.fieldnames
    )

    writer.writeheader()

    for row in records:
        writer.writerow(row)

# JSON Summary
summary = {
    "Target Year": TARGET_YEAR,
    "Total Records": total_records,
    "Incident Count": incident_count,
    "Attack Counts": dict(attack_counter),
    "Total Financial Loss": round(total_loss, 2),
    "Average Financial Loss": round(average_loss, 2)
}

json_file = output_dir / "cybersecurity_summary.json"

with open(json_file, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4)

# Text Report
report_file = output_dir / "cybersecurity_report.txt"

with open(report_file, "w", encoding="utf-8") as report:

    report.write(f"Target Year: {TARGET_YEAR}\n")
    report.write(f"Total Records: {total_records}\n")
    report.write(f"Incident Count: {incident_count}\n")
    report.write(f"Total Financial Loss: {round(total_loss,2)}\n")
    report.write(f"Average Financial Loss: {round(average_loss,2)}\n\n")

    report.write("Attack Counts:\n")

    for attack, count in attack_counter.items():
        report.write(f"{attack}: {count}\n")

    report.write("\nFive Costliest Incidents:\n")

    for incident in top_incidents:
        report.write(
            f"{incident['Country']} | "
            f"{incident['Attack Type']} | "
            f"{incident['Financial Loss (in Million $)']}\n"
        )

    report.write(f"\nFiltered CSV: {csv_file}\n")
    report.write(f"JSON Summary: {json_file}\n")

# Backup JSON
backup_file = backup_dir / json_file.name

shutil.copy2(json_file, backup_file)

# Results
print("Target Year:", TARGET_YEAR)
print("Total Records:", total_records)
print("Incident Count:", incident_count)
print("Attack Counts:", dict(attack_counter))
print("Total Financial Loss:", round(total_loss, 2))
print("Average Financial Loss:", round(average_loss, 2))
print("Top 5 Costliest Incidents:", len(top_incidents))
print("Exported CSV:", csv_file)
print("JSON Summary:", json_file)
print("Backup Exists:", backup_file.exists())
print("Report File:", report_file)
```
