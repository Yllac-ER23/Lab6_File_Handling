```python
from pathlib import Path
import csv
import json
import shutil
from collections import Counter

# Student Parameters
SEED_DIGIT = 6
ID_SUM = 23

RECORDS_TO_PROCESS = 1000 + ID_SUM
TARGET_CLASS = str((SEED_DIGIT % 4) + 2)

data_dir = Path("data")
output_dir = Path("output")
backup_dir = Path("backup")

output_dir.mkdir(exist_ok=True)
backup_dir.mkdir(exist_ok=True)

dataset_path = data_dir / "weblog.csv"

selected_records = []
status_count = Counter()
url_count = Counter()

processed = 0

with open(dataset_path, "r", encoding="utf-8") as file:

    reader = csv.DictReader(file)

    for row in reader:

        if processed >= RECORDS_TO_PROCESS:
            break

        processed += 1

        status = row["Status"]

        if status.startswith(TARGET_CLASS):

            selected_records.append(row)

            status_count[status] += 1
            url_count[row["URL"]] += 1

filtered_csv = output_dir / "filtered_logs.csv"

with open(filtered_csv, "w", newline="", encoding="utf-8") as file:

    writer = csv.DictWriter(
        file,
        fieldnames=["IP", "Time", "URL", "Status"]
    )

    writer.writeheader()

    for row in selected_records:
        writer.writerow(row)

top_5_urls = url_count.most_common(5)

summary = {
    "records_processed": processed,
    "target_status_class": TARGET_CLASS + "xx",
    "records_in_target_class": len(selected_records),
    "count_per_status_code": dict(status_count),
    "five_most_requested_urls": top_5_urls
}

json_file = output_dir / "log_summary.json"

with open(json_file, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4)

backup_file = backup_dir / json_file.name

shutil.copy2(json_file, backup_file)

print("Records Processed:", processed)
print("Target Status Class:", TARGET_CLASS + "xx")
print("Records in Target Class:", len(selected_records))
print("Count per Status Code:", dict(status_count))
print("Top 5 URLs:", top_5_urls)
print("Filtered CSV:", filtered_csv)
print("JSON Summary:", json_file)
print("Backup Exists:", backup_file.exists())
```
